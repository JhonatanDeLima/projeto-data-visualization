```python
import pandas as pd
```

```python
df = pd.read_csv('/content/imigrantes_canada.csv')
```

```python
df
```

```python
df.info()
```

```python
df.set_index('País', inplace=True)
```

```python
anos = list(map(str,range(1980, 2014)))
```

```python
anos
```

```python
brasil = df.loc['Brasil', anos]
```

```python
brasil
```

```python
brasil_dict = {'ano': brasil.index.tolist(), 'imigrantes': brasil.values.tolist()}
dados_brasil = pd.DataFrame(brasil_dict)
```

```python
dados_brasil
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,4))
plt.plot(dados_brasil['ano'], dados_brasil['imigrantes'])
plt.title('Imigração do Brasil para o Canadá')
plt.xlabel('Ano')
plt.ylabel('Número de imigrantes')
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010'])
plt.show()
```

```python
from matplotlib.lines import lineStyles
fig, ax = plt.subplots(figsize=(8,4))
ax.plot(dados_brasil['ano'], dados_brasil['imigrantes'], lw=3, color='g')
ax.set_title('Imigração do Brasil para o Canadá\n1980 a 2013', fontsize=18, loc='left')
ax.set_xlabel('Ano', fontsize=14)
ax.set_ylabel('Número de imigrantes', fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.show()
```

```python
america_sul = df.query('Região=="América do Sul"')
```

```python
america_sul
```

```python
america_sul_sorted = america_sul.sort_values('Total', ascending=True)
```

```python
cores = []
for pais in america_sul_sorted.index:
  if pais == 'Brasil':
    cores.append('green')
  else:
      cores.append('silver')


fig, ax = plt.subplots(figsize=(12,5))
ax.barh(america_sul_sorted.index, america_sul_sorted['Total'], color=cores)

ax.set_title('América do Sul: Brasil foi o quarto país com mais imigrantes\npara o Canadá no período de 1980 a 2013', loc='left', fontsize=18)
ax.set_xlabel('Números de imigrantes', fontsize=14)
ax.set_ylabel('Países', fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)


for i, v in enumerate(america_sul_sorted['Total']):
  ax.text(v + 20, i, str(v), color='black', fontsize=10, ha='left', va='center')

  ax.set_frame_on(False)
  ax.get_xaxis().set_visible(False)
  ax.tick_params(axis='both', which='both', length=0)

  fig.savefig('imigraco_brasil_canada.png', transparent=False, dpi=300,bbox_inches='tight')



plt.show()
```

```python
# Dados da Argentina
argentina = df.loc['Argentina', anos]

argentina_dict = {
    'ano': argentina.index.tolist(),
    'imigrantes': argentina.values.tolist()
}

dados_argentina = pd.DataFrame(argentina_dict)

# Transformando a coluna ano em inteiro
dados_brasil['ano'] = dados_brasil['ano'].astype(int)
dados_argentina['ano'] = dados_argentina['ano'].astype(int)
```

```python
import plotly.graph_objs as go

# Criando a figura
fig = go.Figure()

# Linha do Brasil
fig.add_trace(
    go.Scatter(
        x=[dados_brasil['ano'].iloc[0]],
        y=[dados_brasil['imigrantes'].iloc[0]],
        mode='lines',
        name='Brasil',
        line=dict(width=4, color='green')
    )
)

# Linha da Argentina
fig.add_trace(
    go.Scatter(
        x=[dados_argentina['ano'].iloc[0]],
        y=[dados_argentina['imigrantes'].iloc[0]],
        mode='lines',
        name='Argentina',
        line=dict(width=4, color='blue')
    )
)

# Layout
fig.update_layout(
    title='Imigração do Brasil e Argentina para o Canadá (1980-2013)',
    xaxis=dict(range=[1980, 2013], title='Ano'),
    yaxis=dict(range=[0, 3000], title='Número de Imigrantes'),
    width=1000,
    height=500,
    updatemenus=[
        dict(
            type='buttons',
            showactive=False,
            buttons=[
                dict(
                    label='Play',
                    method='animate',
                    args=[
                        None,
                        {
                            'frame': {'duration': 100, 'redraw': True},
                            'fromcurrent': True
                        }
                    ]
                )
            ]
        )
    ]
)

# Frames da animação
frames = []

for i in range(len(dados_brasil)):
    frames.append(
        go.Frame(
            data=[
                go.Scatter(
                    x=dados_brasil['ano'].iloc[:i+1],
                    y=dados_brasil['imigrantes'].iloc[:i+1]
                ),
                go.Scatter(
                    x=dados_argentina['ano'].iloc[:i+1],
                    y=dados_argentina['imigrantes'].iloc[:i+1]
                )
            ]
        )
    )

fig.frames = frames

fig.show()
```
